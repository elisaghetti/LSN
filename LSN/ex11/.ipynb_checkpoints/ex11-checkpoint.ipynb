## 11.2

**Try to extend the model to obtain a reasonable fit of the following polynomial of order 3:**

$$
f(x)=4-3x-2x^2+3x^3
$$
**for $x \in [-1,1]$.**

I will at first keep the optimizer fixed using Adam, because it is one of the newer available and works well on a variety of problems. I will first work on the number of layers and neurons, aand then once I fixed that try to vary the activation function and loss function. For a fixed number of neurons, (I will try 8,12 and 20)I will consider 2 cases: a "low" number of layers (2) with both "symmetric" and asymmetric ditribution of neurons(like 2 and 10 or 6 and 6) , and a high number of layers in those cases.

```python
#target parameters
a = 3
b=-2
c=-3
d=4
```

```python

np.random.seed(0)
x_train = np.random.uniform(-1, 1, 500)
x_valid = np.random.uniform(-1, 1, 50)
x_valid.sort()
sigma = 0.1
y_target = a*(x_valid**3)+b*(x_valid**2)+c*x_valid+d
plt.grid()
plt.plot(x_valid, y_target, label='target',linestyle='dashed',color='black')

#sigma = 0.10 # noise standard deviation, for the moment it is absent
y_train = np.random.normal(a*(x_train**3)+b*(x_train**2)+c*x_train+d, sigma) # actual measures from which we want to guess regression parameters
y_valid = np.random.normal(a*(x_valid**3)+b*(x_valid**2)+c*x_valid+d, sigma)

plt.scatter(x_valid, y_valid, label=f'sigma ={sigma}')

plt.legend()
plt.title('Validation dataset')

plt.show()
```

```python
s_model_list =[]
a_model_list = []
Nlayers = [2,8]
Nneurons_tot = [16,24,32]

def assign_asymmetric_neurons(model,Nlayers,Nneurons_tot):
    neurons_per_layer = int((Nneurons_tot-Nlayers/2.)/(Nlayers/2))
    for i in range(int(Nlayers/2)) :
        model.add(Dense(1))
        print(1)
        model.add(Dense(neurons_per_layer),activation = 'relu')
        print(neurons_per_layer)
    model.add(Dense(1),activation = 'relu')

def assign_symmetric_neurons(model,Nlayers,Nneurons_tot):
    neurons_per_layer =int( Nneurons_tot/Nlayers)
    for i in range(Nlayers):
        model.add(Dense(neurons_per_layer),activation = 'relu')
    model.add(Dense(1),activation = 'relu')
        
for nn in Nneurons_tot:
    for nl in Nlayers:
        model_s = Sequential()
        model_s.add(Input(shape=(1,)))
        assign_symmetric_neurons(model_s,nl,nn)
        print(f'Using neural network with {nl} layers and {nn} total neurons distributed symmetrically')
        model_s.compile(optimizer='adam', loss='mse', metrics=['mse'])
        model_s.summary()
        s_model_list.append(model_s)
        model_a = Sequential()
        model_a.add(Input(shape=(1,)))
        assign_asymmetric_neurons(model_a,nl,nn)
        print(f'Using neural network with {nl} layers and {nn} total neurons distributed asymmetrically')
        model_a.compile(optimizer='adam', loss='mse', metrics=['mse'])
        model_a.summary()
        a_model_list.append(model_a)
        
```

```python
s_history_list = []
s_predicted_list = []
x= np.linspace(-1,1,100)
for model in s_model_list:
    
    history = model.fit(x=x_train, y=y_train, 
          batch_size=32, epochs=50,
          shuffle=True, # a good idea is to shuffle input before at each epoch
          validation_data=(x_valid, y_valid))
    y = model.predict(x)
    s_predicted_list.append(y)
    s_history_list.append(history.history['loss'])
```

```python
plt.plot(history.history['loss'])
plt.title('Model loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Test'], loc='best')
plt.show()
```
