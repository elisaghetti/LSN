The initial configuration is an Ising system composed of 50 particles, with 50% of the spins equal to $+1/2$ and the remaining 50% equal to $-1/2$. The goal of the exercise was to simulate the system using two different Monte Carlo methods: the **Metropolis algorithm** and the **Gibbs algorithm**.

The Metropolis algorithm was already implemented in the simulator. At each step, it computes an acceptance probability and attempts to flip each spin according to this probability. The acceptance probability corresponds to the Boltzmann weight associated with flipping spin $i$:
$\exp\left(-\beta \Delta E\right)$,
where the energy variation is given by
$\Delta E = 2J s_i (s_{i+1} + s_{i-1})$,
taking into account only the nearest-neighbor interactions.

In contrast, in the Gibbs algorithm all proposed moves are accepted. However, each spin is assigned a value according to a probability distribution that depends on the local configuration of its neighbors. Specifically, the probability of assigning a given spin value is computed from the corresponding Boltzmann weights. The spins are then updated accordingly.

This is how I implemented the Gibbs algorithm:

### INTERNAL ENERGY
$$\frac{U(N,T)}{N} = \frac{\langle H \rangle}{N}
= \frac{1}{N} \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}$$ 

The internal energy computation was already included in the code, and it is computed  by summing over the values of $\Delta E$, defined as I explained before, for all of the particles. 

````python
import numpy as np

#model parameters
points=100
T = np.linspace(0.45,2.2,num=points)
beta = 1/T
J = 1.0
Ns = 50
th = np.tanh(J/T)
thN= th**Ns
ch = 1/th
````

````python
file_metro = 'InternalEnergy_metro.csv'
file_gibbs = 'InternalEnergy_Gibbs.csv'
e = -J*( th + ch*thN )/( 1 + thN ) #model

plot_property(file_metro,file_gibbs,e,'U/N')
````
